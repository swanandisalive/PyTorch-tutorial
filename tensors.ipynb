### Why create random tensors? 
Random tensors are important because many neural networks first start learning by initializing tensors filled with random numbers and adjust those random numbers to better represent the data.

` Start with random numbers -> look at data -> update random numbers -> look at data -> update random numbers `

```python
# Create a random tensor of size (3, 4)
random_tensor = torch.rand(1, 3, 4)
print(random_tensor)
print(random_tensor.ndim)
```

#### Matrix multiplication
Two main ways of performing multiplication in a neural network
1. Element-Wise multiplication
2. Matrix multiplication

There are two main rules to satisfy before doing matrix multiplication:
1. The **inner dimensions** must match:
* `(3, 2) @ (3, 2)` won't work
* `(2, 3) @ (3, 2)` will work

2. The resulting matrix has the shape of the **outer dimensions**:
* `(2, 3) @ (3, 2)` -> `(2, 2)`

```python
# Matrix multiplication
print(torch.matmul(tensor, tensor)) # this method makes matrix multiplication faster as compared to manual matrix multiplication
# this code is less verbose and faster
# matmul gives dot product of matrices
```

### One of the most common erros in deep learning: Shape errors

```python
# Shapes for matrix multiplication
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])
tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])

# torch.mm(tensor_A, tensor_B) is an alias for torch.matmul(a, b)
print(torch.matmul(tensor_A, tensor_B))
```

To fix our tensor shape issues, we can manipulate the shape of one of our tensors using a **transpose**.
A **transpose** switches the axes or dimensions of a given tensor.

```python
tensor_B.T # Transpose of matrix B
print(tensor_B.T)
print(tensor_B.shape)
print(tensor_B.T.shape)

# Retry matrix multiplication of A and B
print(torch.matmul(tensor_A, tensor_B.T).shape)
print(f"Original shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.shape}")
print(f"New shapes: tensor_A = {tensor_A.shape}, tensor_B = {tensor_B.T.shape}")
print(f"Multiplying: {tensor_A.shape} @ {tensor_B.T.shape} <- inner dimensions must match.")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
print(f"\n Output shape: {output.shape}")
```

## Reshaping, Stacking, Squeezing and Unsqueezing tensors

* Reshaping - reshapes an input tensor to a defined shape
* View - Return a view of an input tensor of certain shape but keep the same memory as the original tensor
* Stacking - combine multiple tensors on top of each other (vstack) or side by side(hstack)
* Squeeze - removes all `1` dimensions from a tensor
* Unsqueeze - add a `1` dimension to a target tensor
* Permute - Return a view of the input with dimensions permuted (swapped) in a certain way

```python
import torch 
x = torch.arange(1., 10.)
print(x, x.shape)
```

```python
# Add an extra dimension
#x_reshaped = x.reshape(1, 7)
#x_reshaped = x.reshaped.shape
#print(x_reshaped) # this should give an error

x_reshaped = x.reshape(1, 9)
print(x_reshaped, x_reshaped.shape)
#x_reshaped = x.reshape(9, 1)
#print(x_reshaped, x_reshaped.shape)

# the dimensions specified inside reshape method must be compatible with the dimensions/size of the original matrix.
# as the number of elements insided the matrix are fixed, reshape must not have higher or lesser number of elements than the original matrix
```

```python
# Change the view
z = x.view(1, 9)
print(z)
print(z.shape)
```

```python
# Changing z changes x (because a view of a tensor shares the same memory as the original tensor)
z[:, 0] = 5
print(z)
print(x)
```

```python
# Stack tensor on top of each other
x_stacked = torch.stack([x, x, x, x], dim=0)
print(x_stacked)
```

```python
# torch.squeeze() - removes all single dimensions from a target tensor
x_reshaped.shape
```

```python
print(x_reshaped.squeeze().shape)
```

```python
print(f"Previous tensor: {x_reshaped}")
print(f"Previous shape: {x_reshaped.shape}")

# Remove extra dimensions from x_reshaped
x_squeezed = x_reshaped.squeeze()
print(f"\nNew tensor: {x_squeezed}")
print(f"New shape: {x_squeezed.shape}")
```

```python
# torch.unsqueeze() - adds a single dimension to a target tensor at a specific dim (dimension)
print(f"Previous target: {x_squeezed}")
print(f"Previous shape: {x_squeezed.shape}")

# Add an extra dimension with unsqueeze 
x_unsqueezed = x_squeezed.unsqueeze(dim=1)
print(f"\nNew tensor: {x_unsqueezed}")
print(f"New shape: {x_unsqueezed.shape}")
```

```python
# torch.permute - rearranges the dimensions of a target tensor in a specified order
x_original = torch.rand(size = (224, 224, 3)) # [height width, colour channels]

# Permute the original tensor to rearrange the axis (or dim) order
x_permuted = x_original.permute(2, 0, 1) # shifts axis 0-.1, 1-.2, 2-.0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
```

## Indexing (selecting data from tensors)
Indexing with PyTorch is similar to indexing with numpy

```python
import torch 
x = torch.arange(1, 10).reshape(1, 3, 3)
print(x)
print(x.shape)
```

```python
# Let's index on our new tensor
x[0]
```

```python
# Let's index on the middle bracket (dim=1)
x[0, 0] # or x[0][0]
```

```python
# Let's index on the most inner bracket (last dimension)
print(x[0, 0, 0]) # or x[0][0][0]

print(x[0, 1, 1])
print(x[1, 2, 2])
```

```python
# you can also use ":" to select "all" of a target dimension
x[:, 0]
```

```python
# get all values of 0th and 1st dimensions but only index 1 of 2nd dimension
x[:, :, 1]
```

```python
# Get all values of the 0 dimension but only the 1 index value of 1st and 2nd dimension
print(x[:, 1, 1])
```

```python
# Index on x to return 9
print(x[0, 2, 2])

# Index on x to return 3, 6, 9
print(x[:, :, 2])
```
